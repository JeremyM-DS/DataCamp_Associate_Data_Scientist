Bienvenue à New York, l’une des villes les plus visitées au monde. Pour répondre à la forte demande d’hébergement temporaire, on y trouve de nombreuses annonces Airbnb, pour des séjours allant de quelques nuits à plusieurs mois. Dans ce projet, nous allons examiner de plus près le marché Airbnb new-yorkais en combinant des données provenant de plusieurs types de fichiers comme `.csv`, `.tsv` et `.xlsx`.

Rappelons que les fichiers **CSV**, **TSV** et **Excel** sont trois formats courants pour stocker des données. 
Trois fichiers contenant des données sur les annonces Airbnb de 2019 sont à votre disposition :

**data/airbnb_price.csv**
Il s’agit d’un fichier CSV contenant des informations sur les prix et les emplacements des annonces Airbnb.
- **`listing_id`** : identifiant unique de l’annonce
- **`price`** : prix par nuit en USD
- **`nbhood_full`** : nom de l’arrondissement et du quartier où se situe l’annonce

**data/airbnb_room_type.xlsx**
Il s’agit d’un fichier Excel contenant des informations sur les descriptions des annonces et les types de chambres.
- **`listing_id`** : identifiant unique de l’annonce
- **`description`** : description de l’annonce
- **`room_type`** : Airbnb propose trois types de chambres : chambres partagées, chambres privées et logements/appartements entiers

**data/airbnb_last_review.tsv**
Il s’agit d’un fichier TSV contenant des informations sur les hôtes et les dates d’avis.
- **`listing_id`** : identifiant unique de l’annonce
- **`host_name`** : nom de l’hôte
- **`last_review`** : date du dernier avis laissé sur l’annonce

```python
# Import necessary packages
import pandas as pd
import numpy as np

# Begin coding here ...
# Use as many cells as you like
```

```python
# --- Load the datasets ---
price = pd.read_csv("data/airbnb_price.csv")
room_type = pd.read_excel("data/airbnb_room_type.xlsx")
last_review = pd.read_csv("data/airbnb_last_review.tsv", sep="\t")

# --- Convert last_review dates to datetime ---
last_review["last_review"] = pd.to_datetime(last_review["last_review"], errors="coerce")

# Clean the price column
price["price"] = (
    price["price"]
    .str.replace("dollars", "", regex=False)
    .str.strip()
    .astype(float)
)
```

```python
# --- 1. Oldest and most recent review dates ---
first_reviewed = last_review["last_review"].min()
last_reviewed = last_review["last_review"].max()

# --- 2. Number of private rooms ---
# Nettoyer la colonne room_type
room_type["room_type_clean"] = (
    room_type["room_type"]
    .astype(str)          # au cas où il y ait des NaN
    .str.strip()          # enlever les espaces au début/fin
    .str.lower()          # uniformiser en minuscules
)

# Compter les chambres privées
nb_private_rooms = (room_type["room_type_clean"] == "private room").sum()
nb_private_rooms

# --- 3. Average price (rounded to 2 decimals) ---
avg_price = round(price["price"].mean(), 2)

# --- 4. Combine results into a single-row DataFrame ---
review_dates = pd.DataFrame({
    "first_reviewed": [first_reviewed],
    "last_reviewed": [last_reviewed],
    "nb_private_rooms": [nb_private_rooms],
    "avg_price": [avg_price]
})

review_dates
```
