Depuis 1901, le prix Nobel compte parmi les distinctions internationales les plus prestigieuses. Chaque année, des prix sont décernés en chimie, littérature, physique, physiologie ou médecine, économie et paix. Au-delà de l'honneur, du prestige et d'une dotation conséquente, les lauréats reçoivent également une médaille d'or à l'effigie d'Alfred Nobel (1833 - 1896), fondateur du prix.

![](Nobel_Prize.png)
