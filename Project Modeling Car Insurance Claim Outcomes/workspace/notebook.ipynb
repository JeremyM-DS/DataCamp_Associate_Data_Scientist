Les compagnies d’assurance investissent beaucoup de temps et d’argent pour optimiser leurs tarifs et estimer avec précision la probabilité qu’un client dépose une demande d’indemnisation. Dans de nombreux pays, l’assurance automobile est obligatoire pour circuler sur la voie publique ; le marché est donc immense !

(`Source: https://www.accenture.com/_acnmedia/pdf-84/accenture-machine-leaning-insurance.pdf`) 

Fort de ce constat, On the Road, un assureur auto, vous sollicite pour construire un modèle capable de prédire si un client fera une demande d’indemnisation pendant la période de couverture. Comme ils disposent de peu d’expertise et d’infrastructures pour déployer et suivre des modèles de Machine Learning, ils vous demandent d’identifier l’unique caractéristique qui permet d’obtenir le modèle le plus performant, mesuré par l’accuracy, afin de démarrer avec un modèle simple en production.

Ils vous ont fourni leurs données clients dans un fichier csv nommé `car_insurance.csv`, ainsi qu’un tableau détaillant ci-dessous les noms de colonnes et leurs descriptions.



## Le jeu de données

| Colonne | Description |
|--------|-------------|
| `id` | Identifiant unique du client |
| `age` | Âge du client : <br> <ul><li>`0` : 16-25</li><li>`1` : 26-39</li><li>`2` : 40-64</li><li>`3` : 65+</li></ul> |
| `gender` | Sexe du client : <br> <ul><li>`0` : Femme</li><li>`1` : Homme</li></ul> |
| `driving_experience` | Années de conduite du client : <br> <ul><li>`0` : 0-9</li><li>`1` : 10-19</li><li>`2` : 20-29</li><li>`3` : 30+</li></ul> |
| `education` | Niveau d’études du client : <br> <ul><li>`0` : Aucune scolarité</li><li>`1` : Lycée</li><li>`2` : Université</li></ul> |
| `income` | Niveau de revenu du client : <br> <ul><li>`0` : Pauvreté</li><li>`1` : Classe ouvrière</li><li>`2` : Classe moyenne</li><li>`3` : Classe aisée</li></ul> |
| `credit_score` | Score de crédit du client (entre zéro et un) |
| `vehicle_ownership` | Statut de propriété du véhicule : <br><ul><li>`0` : N’est pas propriétaire de son véhicule (remboursement en cours)</li><li>`1` : Propriétaire de son véhicule</li></ul> |
| `vehcile_year` | Année d’immatriculation du véhicule : <br><ul><li>`0` : Avant 2015</li><li>`1` : 2015 ou après</li></ul> |
| `married` | Statut marital du client : <br><ul><li>`0` : Non marié</li><li>`1` : Marié</li></ul> |
| `children` | Nombre d’enfants du client |
| `postal_code` | Code postal du client | 
| `annual_mileage` | Nombre de miles parcourus par le client chaque année |
| `vehicle_type` | Type de voiture : <br> <ul><li>`0` : Berline</li><li>`1` : Voiture de sport</li></ul> |
| `speeding_violations` | Nombre total d’infractions pour excès de vitesse reçues par le client | 
| `duis` | Nombre de fois où le client a été contrôlé en conduite sous l’emprise de l’alcool |
| `past_accidents` | Nombre total d’accidents antérieurs impliquant le client |
| `outcome` | Le client a-t-il fait une demande d’indemnisation sur son assurance auto (variable de réponse) : <br><ul><li>`0` : Aucune demande</li><li>`1` : Demande effectuée</li></ul> |

```python
# Import required modules
import pandas as pd
import numpy as np
from statsmodels.formula.api import logit

# Start coding!
```

```python
# Load dataset
car = pd.read_csv("car_insurance.csv")

# Features to test
features = [col for col in car.columns if col not in ["id", "outcome"]]

results = []

for feature in features:
    try:
        # Fit logistic regression
        formula = f"outcome ~ {feature}"
        model = logit(formula=formula, data=car).fit(disp=False)

        # Predict probabilities
        preds = model.predict(car[[feature]])

        # Remove NaN predictions
        valid_idx = preds.notna()
        preds_clean = preds[valid_idx]
        actual_clean = car.loc[valid_idx, "outcome"]

        # Convert probabilities to 0/1
        pred_class = np.round(preds_clean)

        # Manual accuracy
        accuracy = (pred_class == actual_clean).mean()

        results.append((feature, accuracy))

    except Exception as e:
        # If a model fails (rare), skip it
        print(f"Feature {feature} skipped due to error: {e}")

# Select best feature
best_feature, best_accuracy = max(results, key=lambda x: x[1])

# Build output DataFrame
best_feature_df = pd.DataFrame({
    "best_feature": [best_feature],
    "best_accuracy": [best_accuracy]
})

best_feature_df
```
